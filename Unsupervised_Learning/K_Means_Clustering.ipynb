```python
import numpy as np

# -----------------------------
# Dataset
# -----------------------------
X = np.array([
    [3, 4],
    [6, 7],
    [4, 5],
    [5, 7],
    [2, 6]
], dtype=float)

K = 2
max_iters = 10
```

```python
# -----------------------------
# Initial Centroids
# (Choosing first two points)
# -----------------------------
centroids = np.array([
    [3, 4],
    [4, 5]
], dtype=float)

print("=" * 60)
print("Initial Centroids")
for i, c in enumerate(centroids):
    print(f"C{i+1} = {c}")

print("=" * 60)
```

```python
import numpy as np

# =============================
# K-Means
# =============================

for iteration in range(max_iters):

    print(f"\nITERATION {iteration+1}")
    print("-" * 60)

    clusters = []

    # -------------------------
    # Assign each point
    # -------------------------
    for point in X:

        distances = []

        for centroid in centroids:
            d = np.sqrt(np.sum((point - centroid) ** 2))
            distances.append(d)

        cluster = np.argmin(distances)
        clusters.append(cluster)

        print(f"Point {point}")
        print(f"   Distance to C1 = {distances[0]:.3f}")
        print(f"   Distance to C2 = {distances[1]:.3f}")
        print(f"   --> Assigned to Cluster {cluster+1}")
        print()

    clusters = np.array(clusters)
    # -------------------------
    # Print Cluster Members
    # -------------------------
    print("-" * 60)

    for i in range(K):
        print(f"Cluster {i+1}")

        cluster_points = X[clusters == i]

        for p in cluster_points:
            print(p)

        print()

    # -------------------------
    # Compute New Centroids
    # -------------------------
    new_centroids = []

    print("-" * 60)
    print("New Centroids")

    for i in range(K):

        cluster_points = X[clusters == i]

        if len(cluster_points) == 0:
            new_centroids.append(centroids[i])
        else:
            new_centroids.append(cluster_points.mean(axis=0))

        print(f"C{i+1} = {new_centroids[-1]}")

    new_centroids = np.array(new_centroids)

    print("-" * 60)

    # -------------------------
    # Check Convergence
    # -------------------------
    if np.allclose(centroids, new_centroids):

        print("Centroids did not change.")
        print("Algorithm Converged!")
        break

    centroids = new_centroids
```

```python

# =============================
# Final Result
# =============================

print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)

for i in range(K):
    print(f"Final Centroid C{i+1} = {centroids[i]}")

print()

for i in range(K):

    print(f"Cluster {i+1}")

    for p in X[clusters == i]:
        print(p)

    print()
```
