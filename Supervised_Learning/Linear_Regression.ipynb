```python
# Hours studied
X = [1, 2, 3, 4, 5]

# Marks
y = [45, 50, 55, 65, 70]
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```

```python
n = len(X)
```

```python
sum_x = sum(X)
sum_y = sum(y)

sum_x2 = 0

for x in X:
    sum_x2 += x ** 2


sum_xy = 0

for i in range(n):
    sum_xy += X[i] * y[i]
```

```python
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)
```

```python
b = (sum_y - m * sum_x) / n
```

```python
hours = 6

prediction = m * hours + b
print(prediction)
```

```python
import matplotlib.pyplot as plt

# Original data
plt.scatter(X, y, label="Training Data")

# Regression line
y_line = []

for x in X:
    y_line.append(m * x + b)

plt.plot(X, y_line)

plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```
