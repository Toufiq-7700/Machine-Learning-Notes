```python
import numpy as np
```

```python
# City Names

cities = ["BA","FI","MI","NA","RM","TO"]
```

```python
# Distance Matrix

distance_matrix = np.array([
    [0,662,877,255,412,996],
    [662,0,295,468,268,400],
    [877,295,0,754,564,138],
    [255,468,754,0,219,869],
    [412,268,564,219,0,669],
    [996,400,138,869,669,0]
])
```

```python
clusters = [[city] for city in cities]

print(clusters)
```

```python
def cluster_distance(cluster1, cluster2):

    min_distance = float('inf')

    for city1 in cluster1:

        for city2 in cluster2:

            i = cities.index(city1)
            j = cities.index(city2)

            if distance_matrix[i][j] < min_distance:
                min_distance = distance_matrix[i][j]

    return min_distance
```

```python
step = 1

while len(clusters) > 1:

    min_dist = float('inf')

    pair = (0,1)

    # Find closest clusters

    for i in range(len(clusters)):

        for j in range(i+1,len(clusters)):

            dist = cluster_distance(clusters[i],clusters[j])

            if dist < min_dist:

                min_dist = dist
                pair = (i,j)

    print("-----------------------------------")
    print("Step",step)

    print("Merge")

    print(clusters[pair[0]],"and",clusters[pair[1]])

    print("Distance =",min_dist)

    # Merge

    new_cluster = clusters[pair[0]] + clusters[pair[1]]

    clusters.pop(pair[1])
    clusters.pop(pair[0])

    clusters.append(new_cluster)

    print("Clusters =",clusters)

    step += 1
```
